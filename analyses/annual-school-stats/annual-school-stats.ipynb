```python
import pandas as pd
```

```python
df = pd \
    .read_csv( '../../data/lottery/merged/sfusd-lottery-all.csv' ) \
    .merge( pd.read_csv( '../../data/lottery/merged/sfusd-program-code-mapping.csv' ) ) \
    .merge( pd.read_csv( '../../data/lottery/merged/sfusd-school-id-mapping.csv' ) )

elementary_count = df \
    [lambda _df: _df['grade'] == 'K'] \
    [lambda _df: _df['ay'] == '24-25'] \
    ['school_name'] \
    .nunique()
school_count = df \
    [lambda _df: _df['ay'] == '24-25'] \
    ['school_name'] \
    .nunique()

print( f"""
There are {elementary_count} elementary schools.
There are {school_count} schools in total.
""".strip() )
```
