```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv( 'rai_composite_scores.csv' )

extended_elementary_schnames = '''
Buena Vista/ Horace Mann K-8
Chinese Immersion School at DeAvila
Lawton Alternative
Malcolm X Academy
Mission Education Center
Tenderloin Community
San Francisco Public Montessori
San Francisco Community Alternative
'''.strip().split( '\n' )

merging_schnames = '''
El Dorado Elementary
Milk (Harvey) Civil Rights Elementary
[[june jordan]]
Malcolm X Academy
San Francisco Community Alternative
San Francisco Public Montessori
Spring Valley Elementary
[[the academy]]
'''.strip().split( '\n' )

closing_schnames = '''
Parker (Jean) Elementary
Sutro Elementary
Yick Wo Elementary
'''.strip().split( '\n' )

print( df.columns )
for col in df.columns:
    if col == 'schname':
        continue

    df[col] = df[col].astype( float )



df = df \
    .sort_values( 'scaled_comm_comp_score', ascending = False ) \
    .reset_index( drop = True ) \
    .assign( closure_eligible = lambda _df: _df.index >= np.median( _df.index ) ) \
    [lambda _df: _df.schname.isin( extended_elementary_schnames ) | _df.schname.str.contains( 'Elementary' )] \
    .assign(
        was_closing = lambda _df: _df.schname.isin( closing_schnames ),
        was_merging = lambda _df: _df.schname.isin( merging_schnames )
    ) \
    .rename( columns = { 'schname': 'School', 'scaled_comm_comp_score': 'RAI score' } ) \
    [['School','RAI score','closure_eligible','was_closing','was_merging']]
df.to_csv( 'rai-elementary-summary.csv', index = False )
display( df )
```
