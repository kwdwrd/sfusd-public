```python
import pandas as pd
```

```python
school_names_file = '../../data/lottery/merged/sfusd-school-names.csv'
rpes_id = pd.read_csv( school_names_file )[lambda _df: _df.school_name.str.contains( 'Parks' )].iloc[0].school_id

program_codes_file = '../../data/lottery/merged/sfusd-program-codes.csv'
rpes_program_codes = pd.read_csv( program_codes_file )[lambda _df: _df.school_id == rpes_id]
```

```python
lottery_file = '../../data/lottery/merged/sfusd-lottery-merged.csv'
pd \
    .read_csv( lottery_file ) \
    [lambda _df: _df.school_id == rpes_id] \
    [['application_year','program_code','grade','requests']] \
    .groupby( [ 'application_year', 'grade' ] ) \
    .sum() \
    .reset_index()
```
