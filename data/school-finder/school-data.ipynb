```python
J = json.loads( school_data_str.replace( '\u0022', '"' ) )
import re

elementaries = []
for s in J:
    # print( s['type'] )
    # if re.search( r'Rosa', s['title'] ):
    if not re.search( r'Elementary', s['type'] ):
        continue

    elementaries.append( {
        'name': s['title'],
        'type': s['type'],
        'sped': s['special_ed_programs'],
        'enrichment': s['field_arts_and_enrichment'],
        'sports': s['field_athletics'],
        'language': s['field_language_program']
    } )



def field_to_set ( f ):
    r = re.findall( r'>([^<]+)', f )
    if r is None:
        return set( [] )
    
    return set( r )



for ix, s in enumerate( elementaries ):
    s['sped'] = re.sub( r'<li>Resource Specialist Program Services<\\\/li>', '', s['sped'] )
    for field in [ 'sped', 'enrichment', 'language', 'sports' ]:
        s[field] = field_to_set( s[field] )

    s['name'] = re.search( r'>([^<]+)<\\\/a>', s['name'] ).group( 1 )
    if len( s['sports'] ) > 0:
        print( f"{s['name']}: {s['sports']}" )

    elementaries[ix] = s



language_count = 0
sports_count = 0
sped_count = 0
enrichment = set()
languages = set()
for s in elementaries:
    language_count += len( s['language'] )
    sports_count += len( s['sports'] )
    sped_count += len( s['sped'] )

    languages = languages.union( s['language'] )
    enrichment = enrichment.union( s['enrichment'] )


print( f'''
Languages: {language_count}
Sports:    {sports_count}
SPED:      {sped_count}

Language:
{languages}

Enrichment:
{enrichment}
''' )
# elementaries
```
